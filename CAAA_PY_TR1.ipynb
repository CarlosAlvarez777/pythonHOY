```python
import random
```

```python
def crear_diccionario():
    ids = random.sample(range(300000, 400000), 50)
    nombres = [
        'Juan', 'Ana', 'Pedro', 'Luis', 'María', 'Elena', 'Carlos', 'Marta', 'Raúl', 'Lucía',
        'David', 'Sofía', 'Andrés', 'Paula', 'Miguel', 'Laura', 'Fernando', 'Carla', 'Jorge', 'Isabel',
        'Sergio', 'Clara', 'Hugo', 'Irene', 'Rafael', 'Sara', 'Manuel', 'Pilar', 'Enrique', 'Julia',
        'Francisco', 'Beatriz', 'Antonio', 'Eva', 'Alejandro', 'Alicia', 'Gabriel', 'Victoria', 'Diego', 'Nuria',
        'Eduardo', 'Patricia', 'Ricardo', 'Rosa', 'Javier', 'Cristina', 'Álvaro', 'Verónica', 'Rubén', 'Silvia'
    ]
    ap_paterno = [
        'Pérez', 'López', 'García', 'Rodríguez', 'Martínez', 'Hernández', 'González', 'Sánchez', 'Ramírez', 'Torres',
        'Cruz', 'Flores', 'Castro', 'Ramos', 'Jiménez', 'Morales', 'Vargas', 'Silva', 'Reyes', 'Ortiz',
        'Mendoza', 'Guerrero', 'Valdez', 'Navarro', 'Romero', 'Salazar', 'Campos', 'Mora', 'Rangel', 'Acosta',
        'Escobar', 'Molina', 'Rivera', 'Aguilar', 'Vega', 'Soto', 'Cortés', 'Fernández', 'Pacheco', 'Villanueva',
        'Muñoz', 'Ávila', 'Miranda', 'Quintero', 'Castillo', 'Pineda', 'Durán', 'Esquivel', 'Cárdenas', 'Zamora'
    ]
    ap_materno = [
        'López', 'Martínez', 'Hernández', 'González', 'Sánchez', 'Rodríguez', 'Ramírez', 'Torres', 'Cruz', 'Flores',
        'Castro', 'Ramos', 'Jiménez', 'Morales', 'Vargas', 'Silva', 'Reyes', 'Ortiz', 'Mendoza', 'Guerrero',
        'Valdez', 'Navarro', 'Romero', 'Salazar', 'Campos', 'Mora', 'Rangel', 'Acosta', 'Escobar', 'Molina',
        'Rivera', 'Aguilar', 'Vega', 'Soto', 'Cortés', 'Fernández', 'Pacheco', 'Villanueva', 'Muñoz', 'Ávila',
        'Miranda', 'Quintero', 'Castillo', 'Pineda', 'Durán', 'Esquivel', 'Cárdenas', 'Zamora', 'Pérez', 'López'
    ]
    edades = [random.randint(18, 30) for _ in range(50)]
    sexos = ['Masculino', 'Femenino'] * 25
    diccionario = {}
    for i in range(len(ids)):
        diccionario[ids[i]] = {
            'Nombre': nombres[i],
            'Apellido Paterno': ap_paterno[i],
            'Apellido Materno': ap_materno[i],
            'Edad': edades[i],
            'Sexo': sexos[i]
        }
    return diccionario
```

```python
def imprimir_diccionario(diccionario):
    print("\n--- Información de los alumnos ---")
    for id_alumno, datos in diccionario.items():
        print(f"\nID: {id_alumno}")
        for clave, valor in datos.items():
            print(f"{clave}: {valor}")
```

```python
def menu():
    diccionario = {}
    while True:
        print("\nMenú:")
        print("1.- Crear Diccionario")
        print("2.- Imprimir Diccionario")
        print("0.- Salir")
        opcion = input("Selecciona una opción: ")
        if opcion == "1":
            diccionario = crear_diccionario()
        elif opcion == "2":
            if diccionario:
                imprimir_diccionario(diccionario)
            else:
                print("El diccionario está vacío. Crea un diccionario primero.")
        elif opcion == "0":
            print("Saliendo del programa.")
            break
        else:
            print("Opción no válida. Inténtalo de nuevo.")
```

```python
menu()
```
